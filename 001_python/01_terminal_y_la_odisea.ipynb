## 1. Terminal, shell y PowerShell

La **terminal** es la ventana en la que escribimos instrucciones.

La **shell** es el programa que interpreta esas instrucciones.

En Windows utilizaremos **PowerShell** como shell.

Cuando se abre PowerShell aparece una línea parecida a esta:

```text
PS C:\Users\nombre>
```

Esa línea es el **prompt**. Indica que PowerShell está listo para recibir una instrucción. El prompt no se copia: escribimos después del símbolo `>`.

### Anatomía de un comando

Observá este ejemplo:

```powershell
Get-Content odisea.html -Head 10
```

- `Get-Content` es el comando.
- `odisea.html` es un argumento: indica sobre qué archivo trabajar.
- `-Head 10` es una opción: pide solamente las primeras diez líneas.

> **Práctica de lectura:** antes de ejecutar cada comando, intentá describir en voz alta qué información recibe y qué resultado esperás.


## 2. ¿Dónde estoy?

El **directorio de trabajo** es la carpeta en la que se encuentra actualmente la terminal.

Ejecutá:

```powershell
pwd
```

`pwd` viene de *print working directory*: mostrar el directorio de trabajo.

En PowerShell, `pwd` es una forma breve de `Get-Location`:

```powershell
Get-Location
```

Ambos comandos producen la misma información.

### Preguntas

1. ¿La ruta incluye tu nombre de usuario?
2. ¿Cuál es la carpeta final de la ruta?
3. ¿Qué separador utiliza Windows entre carpetas?


## 3. ¿Qué hay en esta carpeta?

Para listar el contenido del directorio actual usamos:

```powershell
ls
```

En PowerShell, `ls` es una forma breve de `Get-ChildItem`:

```powershell
Get-ChildItem
```

En el resultado pueden aparecer archivos y carpetas. La columna `Mode` ayuda a distinguirlos: una carpeta suele incluir una `d`.

Para mostrar también elementos ocultos:

```powershell
ls -Force
```

> **Buena práctica:** antes de crear, mover o descargar algo, ejecutá `pwd` y `ls`. Es una manera sencilla de evitar trabajar en la carpeta equivocada.


## 4. Crear un proyecto

Vamos a crear una carpeta para el laboratorio.

```powershell
mkdir laboratorio_pln
```

`mkdir` viene de *make directory*: crear un directorio.

Comprobá que la carpeta exista:

```powershell
ls
```

Entrá en ella:

```powershell
cd laboratorio_pln
```

`cd` viene de *change directory*: cambiar de directorio.

Verificá la nueva ubicación:

```powershell
pwd
```

Ahora creá una carpeta para los datos:

```powershell
mkdir datos
```

Creá otra carpeta que utilizaremos más adelante para el código:

```powershell
mkdir src
```

Inspeccioná la estructura:

```powershell
ls
```

La estructura debería ser semejante a esta:

```text
laboratorio_pln/
├── datos/
└── src/
```


## 5. Rutas relativas y absolutas

Una **ruta absoluta** describe una ubicación desde el comienzo del sistema de archivos:

```text
C:\Users\nombre\laboratorio_pln\datos
```

Una **ruta relativa** describe una ubicación desde el directorio actual:

```text
datos
```

Entrá en `datos` usando una ruta relativa:

```powershell
cd datos
```

Volvé a la carpeta superior:

```powershell
cd ..
```

Los dos puntos `..` significan “la carpeta superior”.

Volvé a entrar:

```powershell
cd datos
```

### Otros atajos útiles

Ir a la carpeta personal:

```powershell
cd ~
```

Limpiar la pantalla sin borrar archivos ni historial:

```powershell
clear
```

> **Autocompletado:** comenzá a escribir el nombre de una carpeta y presioná `Tab`. PowerShell intentará completar el nombre. Esto ahorra tiempo y reduce errores de escritura.

### Rutas con espacios

Cuando una ruta contiene espacios, la encerramos entre comillas:

```powershell
cd "C:\Users\nombre\Mis documentos"
```


## 6. Descargar *La Odisea*

Vamos a descargar la edición HTML de *La Odisea* publicada por Project Gutenberg.

Página del libro:

<https://www.gutenberg.org/ebooks/58221>

Archivo HTML que utilizaremos:

<https://www.gutenberg.org/files/58221/58221-h/58221-h.htm>

Asegurate de estar dentro de `laboratorio_pln\datos`:

```powershell
pwd
```

Descargá el archivo:

```powershell
curl.exe -L "https://www.gutenberg.org/files/58221/58221-h/58221-h.htm" -o 58221-h.htm
```

En este comando:

- `curl.exe` transfiere datos desde una URL;
- `-L` permite seguir redirecciones;
- la URL identifica el recurso;
- `-o` indica el nombre del archivo de salida;
- `58221-h.htm` es el nombre inicial del archivo local.

Comprobá la descarga:

```powershell
ls
```

> Si PowerShell informa un error, no repitas el comando a ciegas. Leé el mensaje, comprobá la conexión, revisá la URL y ejecutá `pwd` para confirmar dónde estás.


## 7. Renombrar, copiar y describir archivos

El nombre `58221-h.htm` no explica bien el contenido. Renombralo:

```powershell
mv 58221-h.htm odisea.html
```

`mv` viene de *move*. Se utiliza tanto para mover como para renombrar.

Comprobá el cambio:

```powershell
ls
```

Creá una copia de respaldo:

```powershell
cp odisea.html odisea_original.html
```

`cp` viene de *copy*.

Volvé a listar:

```powershell
ls
```

Obtené información del archivo de trabajo:

```powershell
Get-Item odisea.html
```

Mostrá solamente el nombre y el tamaño en bytes:

```powershell
Get-Item odisea.html | Select-Object Name, Length
```

El símbolo `|` se llama **pipe**. Envía el resultado del comando de la izquierda al comando de la derecha.

> No necesitás dominar los pipes hoy. Lo importante es reconocer que varios comandos pequeños pueden conectarse para construir una operación mayor.


## 8. Inspeccionar el archivo

Una página web no se descarga como una captura de pantalla. Se descarga como código HTML.

Mostrá las primeras veinte líneas:

```powershell
Get-Content odisea.html -Encoding utf8 -Head 20
```

Mostrá las últimas diez líneas:

```powershell
Get-Content odisea.html -Encoding utf8 -Tail 10
```

Contá líneas, palabras y caracteres:

```powershell
Get-Content odisea.html -Encoding utf8 | Measure-Object -Line -Word -Character
```

## 9. Buscar dentro del HTML

Buscá la expresión `CANTO PRIMERO`:

```powershell
Select-String -Path odisea.html -Pattern "CANTO PRIMERO"
```

Buscá el nombre `Ulises`:

```powershell
Select-String -Path odisea.html -Pattern "Ulises"
```

Hacé una búsqueda sin distinguir mayúsculas de minúsculas:

```powershell
Select-String -Path odisea.html -Pattern "ulises" -CaseSensitive:$false
```

Contá cuántas líneas contienen `Ulises`:

```powershell
Select-String -Path odisea.html -Pattern "Ulises" | Measure-Object
```

Este número no es necesariamente la frecuencia exacta de la palabra. Una misma línea puede contenerla varias veces. La diferencia entre **líneas coincidentes** y **apariciones** será importante cuando hagamos análisis textual.

## 11. Mapa de comandos

| Necesidad | Forma breve | Forma explícita en PowerShell |
| --- | --- | --- |
| Mostrar la ubicación | `pwd` | `Get-Location` |
| Listar contenido | `ls` | `Get-ChildItem` |
| Cambiar de carpeta | `cd` | `Set-Location` |
| Crear una carpeta | `mkdir` | `New-Item -ItemType Directory` |
| Mover o renombrar | `mv` | `Move-Item` |
| Copiar | `cp` | `Copy-Item` |
| Mostrar un archivo | — | `Get-Content` |
| Buscar texto | — | `Select-String` |
| Medir contenido | — | `Measure-Object` |
| Limpiar la pantalla | `clear` | `Clear-Host` |

Los alias son cómodos y se parecen a los comandos usados en macOS y Linux. Los nombres explícitos ayudan a leer scripts de PowerShell y a consultar la documentación.
