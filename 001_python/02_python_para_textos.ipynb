## 6. Indexación y slicing

````python
palabra = "Penélope"

primer_caracter = palabra[0]

ultimo_caracter = palabra[-1]

primeras_tres_letras = palabra[0:3]

print(primer_caracter)

print(ultimo_caracter)

print(primeras_tres_letras)
````

## 7. Métodos de strings

Un método es una operación asociada a un objeto. Se escribe después del string o de la variable, separado por un punto.

Vamos a conservar cada estado de la transformación en una variable diferente. Así podemos inspeccionar y explicar cada paso.


````python
texto_original = "  Ulises vuelve a Ítaca. Ulises reconoce su hogar.  "

texto_sin_espacios_externos = texto_original.strip()

texto_en_minusculas = texto_sin_espacios_externos.lower()

texto_sin_puntos = texto_en_minusculas.replace(".", "")

palabras = texto_sin_puntos.split()

print(texto_original)

print(texto_sin_espacios_externos)

print(texto_en_minusculas)

print(texto_sin_puntos)

print(palabras)
````

Los strings son **inmutables**: los métodos no modifican el valor original. Devuelven un string nuevo. Por eso guardamos cada resultado en otra variable.

Observá también que `.split()` cambia el tipo del dato: recibe un string y devuelve una lista.


````python
print(texto_original)

print(type(texto_original))

print(type(palabras))
````

## 8. Listas: una colección ordenada

Una lista agrupa varios elementos y conserva su orden. `.split()` nos dio una lista de palabras. También podemos escribir una lista directamente.


````python
personajes = [
    "Ulises",
    "Penélope",
    "Telémaco",
    "Atenea",
]

primer_personaje = personajes[0]

ultimo_personaje = personajes[-1]

cantidad_de_personajes = len(personajes)

print(primer_personaje)

print(ultimo_personaje)

print(cantidad_de_personajes)
````

El operador `in` pregunta si un elemento pertenece a una colección. El resultado es un booleano.


````python
aparece_ulises = "Ulises" in personajes

aparece_homero = "Homero" in personajes

print(aparece_ulises)

print(aparece_homero)
````

## 9. Bucles `for`: repetir una operación

````python
for personaje_actual in personajes:
    mensaje_personaje = f"Personaje: {personaje_actual}"

    print(mensaje_personaje)
````

## 10. Condicionales: tomar decisiones

`if` ejecuta un bloque solamente cuando una condición es verdadera. `else` permite indicar qué hacer en el caso contrario.


````python
corpus_en_minusculas = corpus_breve.lower()

corpus_sin_puntos = corpus_en_minusculas.replace(".", "")

palabras_del_corpus = corpus_sin_puntos.split()

for palabra_actual in palabras_del_corpus:
    longitud_actual = len(palabra_actual)

    if longitud_actual >= 7:
        mensaje_longitud = f"{palabra_actual} tiene siete caracteres o más."

        print(mensaje_longitud)
````

> Cada transformación quedó en una variable distinta. Si el resultado no fuera el esperado, podríamos imprimir cada estado y localizar el paso problemático.


## 11. Diccionarios: claves y valores

Un diccionario relaciona claves con valores. Para un contador de palabras, cada palabra será una clave y su frecuencia será el valor.


````python
frecuencias_de_ejemplo = {
    "ulises": 3,
    "ítaca": 2,
    "penélope": 1,
}

frecuencia_de_ulises = frecuencias_de_ejemplo["ulises"]

print(frecuencia_de_ulises)
````

### Construir frecuencias explícitamente

Primero escribiremos todas las decisiones. Esta versión es más larga, pero permite ver el algoritmo.


````python
texto_para_contar = corpus_breve.strip()

texto_para_contar = texto_para_contar.lower()

texto_para_contar = texto_para_contar.replace(".", "")

palabras_para_contar = texto_para_contar.split()

frecuencias = {}

for palabra_actual in palabras_para_contar:
    if palabra_actual in frecuencias:
        frecuencia_actual = frecuencias[palabra_actual]

        frecuencia_nueva = frecuencia_actual + 1

        frecuencias[palabra_actual] = frecuencia_nueva
    else:
        frecuencias[palabra_actual] = 1

print(frecuencias)
````

### La misma actualización con `.get()`

`diccionario.get(clave, valor_predeterminado)` devuelve el valor existente. Si la clave todavía no existe, devuelve el valor predeterminado. No es necesario memorizar esta forma hoy; sirve para reconocerla en código ajeno.


````python
frecuencias_con_get = {}

for palabra_actual in palabras_para_contar:
    frecuencia_actual = frecuencias_con_get.get(palabra_actual, 0)

    frecuencia_nueva = frecuencia_actual + 1

    frecuencias_con_get[palabra_actual] = frecuencia_nueva

print(frecuencias_con_get)
````

## 12. Informar resultados con f-strings


````python
cantidad_total = len(palabras_para_contar)

cantidad_diferente = len(frecuencias)

informe_total = f"Palabras totales: {cantidad_total}"

informe_diferentes = f"Palabras diferentes: {cantidad_diferente}"

print(informe_total)

print(informe_diferentes)

print(frecuencias)
````

## 13. Funciones: dar nombre a una transformación

Una función agrupa instrucciones relacionadas. Puede recibir datos mediante parámetros y devolver un resultado con `return`.

La docstring explica el propósito de la función.


````python
def normalizar_texto(texto_original):
    """Prepara un texto para un primer análisis exploratorio."""

    texto_sin_espacios = texto_original.strip()

    texto_en_minusculas = texto_sin_espacios.lower()

    texto_sin_puntos = texto_en_minusculas.replace(".", "")

    texto_sin_comas = texto_sin_puntos.replace(",", "")

    return texto_sin_comas
````

````python
texto_de_prueba = "  Ulises viaja, Penélope espera.  "

texto_normalizado = normalizar_texto(texto_de_prueba)

print(texto_normalizado)
````
