Antes de escribir una línea de scraping confirmamos que la computadora tiene todo

## 1. Qué Python está usando esta notebook

```python
import sys

print("Versión:  ", sys.version.split()[0])
print("Ejecutable:", sys.executable)
```

La versión tiene que empezar con **3.12** y el ejecutable tiene que pasar por una
carpeta `.venv`.

Si dice otra cosa, VS Code está usando otro Python:
`Ctrl+Shift+P` → `Python: Select Interpreter` → elegí el que dice `.venv`.

## 2. Están las cuatro librerías

```python
import bs4
import matplotlib
import pandas
import requests
import trafilatura

print("requests      ", requests.__version__)
print("beautifulsoup4", bs4.__version__)
print("trafilatura   ", trafilatura.__version__)
print("pandas        ", pandas.__version__)
print("matplotlib    ", matplotlib.__version__)
```

## 3. Puedo leer y escribir archivos

Todo lo que produzcamos hoy va a la carpeta `datos/`.

```python
from pathlib import Path

DATOS = Path("datos")
DATOS.mkdir(exist_ok=True)

(DATOS / "prueba.txt").write_text("hola", encoding="utf-8")
leido = (DATOS / "prueba.txt").read_text(encoding="utf-8")

print("Carpeta de trabajo:", Path.cwd())
print("Puedo escribir:    ", leido == "hola")
print("Tengo stopwords:   ", (DATOS / "stopwords_es.txt").exists())
```

## 4. Tengo salida a internet

```python
import requests

respuesta = requests.get("https://www.gutenberg.org", timeout=20)
print("Estado:", respuesta.status_code, "(esperamos 200)")
```

## Si algo falló

| Lo que dice | Lo que pasa | Qué hacer |
|---|---|---|
| `ModuleNotFoundError` | Falta una librería o VS Code usa otro Python | `uv pip install -r requirements.txt` y volvé a elegir el intérprete `.venv` |
| La versión no es 3.12 | VS Code está usando el Python del sistema | `Ctrl+Shift+P` → `Python: Select Interpreter` |
| `Puedo escribir: False` | La notebook se abrió desde otra carpeta | Abrí VS Code **sobre la carpeta de la materia**, no sobre el archivo suelto |
| Falla el `requests.get` | Sin red, o un proxy/firewall en el medio | Avisá en el chat del Meet |

**Levantá la mano en el Meet si algo de esto no da verde.** No sigas.
