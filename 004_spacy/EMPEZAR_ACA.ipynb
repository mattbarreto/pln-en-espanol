## 1. Dónde quedamos

En text mining aprendiste a convertir textos en números y después a desconfiar de esos
números. Tres cosas quedaron dichas:

- Convertir texto en números **destruye información**, y se hace a propósito.
- Un conteo crudo compara tamaños de archivo, no lo que querés comparar.
- Una imagen puede sostener una afirmación o taparla.

La herramienta era la **bolsa de palabras**: cortar el texto, contar, vectorizar.
Funciona, y tiene un techo.

## 2. Contra qué choca la bolsa de palabras

Tres problemas, y los tres se ven con Python solo. No hace falta ninguna librería para
comprobarlos.

### El primero: la misma palabra no es la misma cosa

```python
frase_1 = "Ayer vino mi hermana."
frase_2 = "Serví un vino tinto."

for frase in [frase_1, frase_2]:
    palabras = frase.lower().replace(".", "").split()
    print(palabras)
```

En las dos listas aparece `vino`. Para un contador es la misma entrada, y suma dos.

Pero en la primera es un **verbo** —de *venir*— y en la segunda un **sustantivo**. Contar
las dos juntas no es un error de redondeo: es contar dos cosas distintas como si fueran
una.

### El segundo: la misma cosa no es la misma palabra

```python
formas_de_ser = ["soy", "sos", "es", "era", "fui", "seremos", "siendo"]

for forma in formas_de_ser:
    print(forma)

print()
print("entradas distintas en un contador:", len(formas_de_ser))
```

Siete entradas para un contador. Una sola palabra para cualquier lector.

El español conjuga mucho, y eso reparte la frecuencia de un mismo verbo entre decenas de
formas. Ninguna llega a aparecer en el top de nada, aunque juntas sean lo más frecuente
del corpus.

### El tercero: el orden no está

```python
frase_a = "La vecina saludó a la portera."
frase_b = "La portera saludó a la vecina."

palabras_a = sorted(frase_a.lower().replace(".", "").split())
palabras_b = sorted(frase_b.lower().replace(".", "").split())

print("a:", palabras_a)
print("b:", palabras_b)
print()
print("¿misma bolsa de palabras?", palabras_a == palabras_b)
```

Las dos frases tienen exactamente la misma bolsa de palabras, y dicen cosas opuestas.

Quién saluda y quién es saludado no está en las palabras: está en **cómo se relacionan**.
La bolsa tira esa información en el primer paso.

> Los tres problemas son el mismo problema con tres caras: la bolsa de palabras trata al
> texto como un conjunto de fichas sueltas, y el lenguaje no es eso.

## 3. Qué hace spaCy

spaCy no cuenta palabras. **Anota** el texto: le agrega capas de información encima, y
después vos contás sobre esas capas.

Es una sola librería para toda la unidad.

| Lo que agrega | Qué resuelve | Dónde lo vemos |
|---|---|---|
| **Categoría gramatical** | *vino* verbo y *vino* sustantivo dejan de ser lo mismo | `02` |
| **Lema** | Las siete formas de *ser* se cuentan juntas | `02` |
| **Morfología** | Tiempo, modo, persona y número de cada verbo | `02` |
| **Sintaxis** | Quién le hace qué a quién | `03` |
| **Entidades** | Qué nombres propios hay y de qué tipo | `03` |

Y una cuarta cosa, que es la que ocupa la mitad de la unidad: **spaCy se equivoca**, y se
equivoca de maneras que se pueden medir, auditar y corregir. El `04` mide los errores y
el `05` los corrige.
