## ◈ Microglosario

| Término | Qué es en lenguaje llano |
|---|---|
| **audio** | Un archivo de sonido: MP3, WAV, M4A, OGG. Para la computadora, una larga lista de números |
| **`yt-dlp`** | Un programa que baja video y audio de plataformas. Se usa desde la terminal o desde Python |
| **FFmpeg** | La navaja suiza del audio y el video. Convierte entre formatos. No es una librería de Python: es un programa aparte |
| **procedencia** | De dónde salió cada archivo: URL, autoría, fecha, licencia. Sin eso no hay corpus |
| **licencia** | Lo que el autor permite hacer con su material. No es lo mismo escuchar que redistribuir |
| **corpus de audio** | Una colección de audios elegidos con un criterio, con su procedencia anotada al lado |

## Por qué ir a buscar audio

En la unidad 2 armaste un corpus bajando páginas web. Funcionó porque ese material ya
estaba escrito: alguien lo tipeó, lo publicó y quedó ahí.

Buena parte del lenguaje no pasa por ahí. Entrevistas, clases, debates parlamentarios,
podcasts, testimonios, notas de voz: eso se produce hablando y queda registrado como
sonido. Si tu pipeline solo lee HTML, ese material no existe para vos.

**El problema técnico es siempre el mismo**, y por eso vale la pena aprenderlo una vez:

```text
una fuente cualquiera → un archivo de audio → Whisper → texto
```

Lo que cambia entre una entrevista de YouTube y un audio de WhatsApp no es el
procesamiento. Es cómo conseguís el archivo. Todo lo demás es idéntico.

## De dónde puede salir el audio

| Tipo de fuente | Ejemplos concretos | Cómo se obtiene |
|---|---|---|
| Plataformas de video | YouTube, Vimeo, Twitch, TikTok, Instagram | `yt-dlp` |
| Podcasts y audio | iVoox, SoundCloud, feeds RSS con `.mp3` directo | `yt-dlp`, descarga directa |
| Repositorios institucionales | Archivos orales de bibliotecas, historia oral, Archivo de la Palabra | descarga directa o scraping |
| Medios | Radios con archivo online, sesiones legislativas, conferencias de prensa | `yt-dlp` o descarga directa |
| Conferencias y educación | Charlas TED, grabaciones de Zoom o Meet, MOOCs | exportación, `yt-dlp` |
| Mensajería | Audios de WhatsApp (`.opus`), Telegram | exportación manual del chat |
| Trabajo de campo propio | Entrevistas que grabás vos, registros etnográficos | grabación directa |

La última fila es la más interesante para un trabajo de investigación, y la única donde
el consentimiento lo gestionás vos directamente.

En este cuaderno usamos `yt-dlp` porque cubre las dos primeras filas, que es de donde va
a salir el material de la mayoría. El resto de la unidad no depende de esta decisión: si
conseguís el audio de otra manera, los cuadernos 01 y 02 funcionan igual.

## Antes de bajar nada: qué podés usar

Esto no es un trámite. Es la parte del cuaderno que más se parece a lo que vas a tener
que decidir después, sola o solo, sin nadie que te diga si está bien.

**Tres cosas distintas que se confunden todo el tiempo:**

| | La pregunta | Ejemplo de respuesta |
|---|---|---|
| **Acceso** | ¿Puedo escucharlo? | Sí, está público en la plataforma |
| **Uso** | ¿Puedo procesarlo para un trabajo de clase? | Casi siempre sí, con fines educativos y sin republicar |
| **Redistribución** | ¿Puedo volver a publicarlo, o publicar su transcripción? | Casi siempre **no**, salvo licencia abierta |

Que algo esté accesible no dice nada sobre las otras dos.

**Para esta unidad, elegí material que cumpla alguna de estas:**

- Licencia Creative Commons o dominio público.
- Contenido institucional público: sesiones legislativas, conferencias de prensa,
  universidades con archivo abierto.
- Material propio, o de alguien que te dio permiso explícito.

**Y una advertencia que no es formal.** Transcribir cambia lo que un audio es. Un audio
hay que escucharlo entero para saber qué dice; una transcripción se busca con Ctrl+F, se
copia, se cita fuera de contexto y se indexa. Convertir voz en texto **vuelve ese material
mucho más fácil de redistribuir**, y eso cae sobre personas que hablaron sin imaginar que
las estaban transcribiendo.

Por eso no vamos a transcribir grabaciones de clase con intervenciones de compañeros, ni
audios privados, ni entrevistas ajenas sin permiso.

## Bajar un audio

`yt_dlp` se configura con un diccionario de opciones. Vamos a leerlo de a una línea,
porque cada opción es una decisión.

Prestá atención a `postprocessors`: eso es lo que llama a FFmpeg. Sin esa parte bajarías
el formato original de la plataforma, que suele ser `.webm` o `.m4a`.

### ✎ Antes de ejecutar

El video que elijas dura unos minutos. El archivo de audio que va a quedar, ¿va a pesar
más o menos que el video original? ¿Por qué?

````python
# Cambiá esta URL por la del material que elegiste vos.
URL_DEL_AUDIO = "https://www.youtube.com/watch?v=yX2tPSjBoeU"

CARPETA_AUDIOS = "../datos/audios"

opciones = {
    "format": "bestaudio/best",          # de todas las pistas, la de mejor audio
    "noplaylist": True,                  # si la URL es de una lista, bajar solo el video
    "ffmpeg_location": carpeta_ffmpeg,   # dónde está FFmpeg
    "outtmpl": CARPETA_AUDIOS + "/%(title)s.%(ext)s",   # cómo nombrar el archivo
    "quiet": True,
}
````

`postprocessors` es una lista de cosas que se hacen **después** de bajar. Acá hay una
sola: extraer el audio y convertirlo a MP3.

Lo separamos en su propia celda porque es exactamente el paso que necesita FFmpeg, y
conviene que se vea dónde entra.

````python
conversion_a_mp3 = {
    "key": "FFmpegExtractAudio",
    "preferredcodec": "mp3",
    "preferredquality": "192",   # kilobits por segundo: suficiente para voz
}

opciones["postprocessors"] = [conversion_a_mp3]

print("Configuración lista. Vamos a bajar de:", URL_DEL_AUDIO)
````

Y ahora sí, la descarga.

`YoutubeDL` se usa con `with`, igual que `open`: eso garantiza que la conexión se cierre
sola aunque algo falle en el medio. Es el mismo patrón que usaste para abrir archivos en
la unidad 1.

Puede tardar. El aviso aparece cuando terminó.

````python
with yt_dlp.YoutubeDL(opciones) as descargador:
    # download=True baja el archivo; con False solo consultaría los metadatos.
    informacion = descargador.extract_info(URL_DEL_AUDIO, download=True)

titulo = informacion["title"]
duracion_en_segundos = informacion["duration"]
canal = informacion["channel"]

print("Título  :", titulo)
print("Canal   :", canal)
print("Duración:", duracion_en_segundos, "segundos")
````

**Fijate en `informacion`.** `extract_info` no devuelve solo el archivo: devuelve un
diccionario con decenas de campos sobre el material. Eso es lo que nos va a permitir
registrar la procedencia sin tipearla a mano.

Es la misma idea de la unidad 2: un corpus es texto —o audio— **más** procedencia. Un
archivo suelto, sin fecha ni fuente ni autoría, no es un dato: es un residuo digital.

## De un audio a un corpus

Un audio suelto alcanza para probar. Para trabajar hace falta una colección con criterio,
y con la procedencia de cada pieza anotada al lado.

Vamos a armar exactamente la misma estructura que en la unidad 2: un archivo `.jsonl` con
un registro por línea, y un índice `.csv` para poder mirarlo en una planilla.

Primero, la lista. **Poné acá las URLs de tu material**, y elegilas con un criterio
que puedas defender: mismo tema, mismo programa, mismo período, voces que se contrastan.

Tres o cuatro alcanzan para la clase.

````python
URLS_DEL_CORPUS = [
    "https://www.youtube.com/watch?v=yX2tPSjBoeU",
    "https://www.youtube.com/watch?v=ruepxLoEwoo",
    "https://www.youtube.com/watch?v=WqXr0AujesY",
]

# Lo que declarás vos, porque yt-dlp no lo sabe.
LICENCIA_DECLARADA = "revisar en la fuente"

cantidad_de_urls = len(URLS_DEL_CORPUS)
print("URLs a procesar:", cantidad_de_urls)
````

`yt-dlp` devuelve la fecha de publicación pegada, como `20251211`. La unidad 2 guardó
las fechas como `2025-12-11`, y conviene que los dos corpus se lean igual.

Cortar un string en pedazos por posición es lo mismo que hiciste en la unidad 1 con *La
Odisea*.

````python
def fecha_en_formato_iso(fecha_pegada):
    """Convierte '20251211' en '2025-12-11'."""
    anio = fecha_pegada[0:4]
    mes = fecha_pegada[4:6]
    dia = fecha_pegada[6:8]
    return anio + "-" + mes + "-" + dia


print("Ejemplo:", fecha_en_formato_iso("20251211"))
````

Ahora los campos que guardamos de cada audio. Son los mismos seis de la unidad 2, más
tres que solo tienen sentido para audio: cuánto dura, en qué archivo quedó y bajo qué
licencia lo estás usando.

````python
def datos_de_procedencia(url, informacion):
    """Arma el registro de un audio: de dónde salió y qué es."""
    nombre_seguro = sanitizar_nombre_archivo(informacion["title"])
    fecha_publicada = fecha_en_formato_iso(informacion["upload_date"])
    return {
        "url": url,
        "medio": informacion["channel"],
        "titulo": informacion["title"],
        "autoria": informacion["uploader"],
        "fecha_publicacion": fecha_publicada,
        "fecha_captura": datetime.date.today().isoformat(),
        "duracion_segundos": informacion["duration"],
        "licencia": LICENCIA_DECLARADA,
        "archivo": nombre_seguro + ".mp3",
    }
````

Y la que junta las dos cosas: baja el audio y devuelve su ficha.

Es corta a propósito. Todo lo que hace ya lo corriste paso a paso más arriba: esto es lo
mismo, envuelto en un nombre para poder repetirlo.

````python
def bajar_y_registrar(url, opciones_de_descarga):
    """Baja el audio de una URL y devuelve su registro de procedencia."""
    with yt_dlp.YoutubeDL(opciones_de_descarga) as descargador:
        informacion = descargador.extract_info(url, download=True)

    return datos_de_procedencia(url, informacion)
````

La celda de abajo recorre la lista. Un audio por vuelta, y un aviso por cada uno:
bajar varios archivos tarda, y una celda muda durante tres minutos parece colgada.

````python
corpus_de_audio = []

for url in URLS_DEL_CORPUS:
    print("Bajando:", url)
    registro = bajar_y_registrar(url, opciones)
    corpus_de_audio.append(registro)
    print("   listo:", registro["titulo"])

cantidad_descargada = len(corpus_de_audio)
print()
print("Audios en el corpus:", cantidad_descargada)
````

Antes de guardar nada, una comprobación.

`assert` es una afirmación sobre el estado del programa: si es falsa, corta ahí con el
mensaje que le pusiste. Sirve para no descubrir dentro de media hora que el corpus estaba
vacío desde el principio.

````python
assert len(corpus_de_audio) > 0, "El corpus quedó vacío: ninguna URL se pudo bajar."

primer_registro = corpus_de_audio[0]
campos_del_registro = list(primer_registro.keys())

print("Campos que guardamos de cada audio:")
for campo in campos_del_registro:
    print("  -", campo)
````

## Guardar el corpus

Dos archivos, igual que en la unidad 2:

- **`corpus_audio.jsonl`** — un registro JSON por línea. Es el corpus.
- **`corpus_audio_indice.csv`** — lo mismo, para abrir en una planilla y mirarlo de un
  vistazo.

`ensure_ascii=False` es lo que hace que las tildes y las eñes se guarden como tildes y
eñes, y no como `\u00f1`.

````python
import json

ARCHIVO_CORPUS = "../datos/corpus_audio.jsonl"

with open(ARCHIVO_CORPUS, "w", encoding="utf-8") as archivo:
    for registro in corpus_de_audio:
        linea = json.dumps(registro, ensure_ascii=False)
        archivo.write(linea + "\n")

print("Guardado en:", ARCHIVO_CORPUS)
````

El `.jsonl` ya está. Ahora el índice, que es el mismo contenido en forma de tabla.

Un `.jsonl` es cómodo para un programa y molesto para una persona: hay que abrirlo con un
editor y leer JSON. El `.csv` se abre con doble clic en cualquier planilla, y sirve para lo
que más vas a hacer con esto —mirarlo y decidir qué sacar.

````python
import pandas as pd

indice = pd.DataFrame(corpus_de_audio)

# El índice se mira, no se procesa: sacamos la ruta del archivo para que entre en pantalla.
indice_para_mirar = indice.drop(columns=["archivo"])

print("Corpus de audio:")
indice_para_mirar
````

**Mirá la tabla de arriba antes de seguir.** Esa es tu decisión de corpus, escrita:
qué elegiste, de dónde salió y cuándo lo capturaste.

Si algo te llama la atención —una duración rarísima, una autoría vacía, un título que no
es lo que creías— corregilo ahora. Después va a ser un número raro en un gráfico y no vas
a saber de dónde salió.

````python
ARCHIVO_INDICE = "../datos/corpus_audio_indice.csv"

indice.to_csv(ARCHIVO_INDICE, index=False)

print("Índice exportado a:", ARCHIVO_INDICE)
print()
print("Ya tenés el material. Seguí en el cuaderno 01.")
````
