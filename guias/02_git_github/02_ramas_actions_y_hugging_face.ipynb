# Parte 1: Configuración Inicial

Antes de proceder con la escritura de código, resultará necesario presentarse ante Git. Git funciona como un notario; precisa conocer la identidad del autor para certificar los cambios.

### 1.1 Configuración de identidad
Abra su terminal (consola) y ejecute los siguientes dos comandos (reemplace con sus datos):

```bash
git config --global user.name "Su Nombre"
git config --global user.email "su@email.com"
```

### 1.2 Conexión con GitHub
Para subir código a GitHub, se requiere permiso. El método actual implica el uso de un **Personal Access Token (PAT)** o **SSH**.

**Opción Rápida (HTTPS con Token):**
1. Diríjase a GitHub -> Settings -> Developer settings -> Personal access tokens -> Tokens (classic).
2. Genere uno nuevo con permisos de `repo`.
3. Cuando la consola solicite contraseña al realizar un `push`, utilizará este Token, no su contraseña de inicio de sesión.

# Parte 2: Git desde la Consola (Lo Esencial)

Dejaremos de lado las interfaces gráficas momentáneamente. Estos 6 comandos constituyen la base del trabajo diario.

### 2.1 Clonar (Descargar)
Permite traer un repositorio de la nube a la computadora local.
```bash
git clone https://github.com/usuario/repo.git
cd repo
```

### 2.2 Estado (Verificación)
Permite visualizar qué archivos sufrieron modificaciones.
```bash
git status
```

### 2.3 Pull (Actualizar)
**Fundamental antes de comenzar a trabajar.** Descarga los cambios más recientes del repositorio remoto. Evita conflictos cuando se trabaja en equipo.
```bash
git pull origin main
```
> ⚠️ **Regla de oro**: Siempre ejecute `git pull` antes de comenzar a editar código.

### 2.4 Agregar (Preparar)
Indica a Git: "Deseo guardar estos cambios".
```bash
git add .
# El punto significa "todo". Se puede especificar un archivo puntual también.
```

### 2.5 Commit (Guardar)
Guarda los cambios con un mensaje descriptivo. Funciona como un punto de control.
```bash
git commit -m "Mi primer cambio importante"
```

### 2.6 Push (Subir)
Envía los cambios guardados al repositorio remoto en GitHub.
```bash
git push origin main
```

### Resumen del Flujo Típico
```bash
git pull origin main      # 1. Actualizar antes de trabajar
# ... editar archivos ...
git status                # 2. Ver qué cambió
git add .                 # 3. Preparar cambios
git commit -m "mensaje"   # 4. Guardar localmente
git push origin main      # 5. Subir a GitHub
```

# Parte 2.5: Trabajando con Ramas (Branches)

Las ramas permiten trabajar en nuevas funcionalidades sin afectar el código principal. Piense en ellas como líneas de tiempo paralelas.

### ¿Por qué usar ramas?
- **Aislamiento**: Los experimentos no afectan el código estable.
- **Colaboración**: Múltiples personas trabajan simultáneamente sin conflictos.
- **Organización**: Cada funcionalidad tiene su propia rama.

### Comandos Esenciales de Ramas

#### Ver ramas existentes
```bash
git branch          # Lista ramas locales
git branch -a       # Lista todas (locales y remotas)
```

#### Crear una rama nueva
```bash
git branch nombre-de-la-rama
```

#### Cambiar de rama
```bash
git checkout nombre-de-la-rama
# O en versiones modernas de Git:
git switch nombre-de-la-rama
```

#### Crear y cambiar en un solo paso
```bash
git checkout -b nueva-funcionalidad
# Equivalente moderno:
git switch -c nueva-funcionalidad
```

#### Fusionar ramas (Merge)
Una vez completada la funcionalidad, se integra a `main`:
```bash
git checkout main              # Volver a main
git pull origin main           # Actualizar main
git merge nueva-funcionalidad  # Fusionar los cambios
git push origin main           # Subir la fusión
```

#### Eliminar una rama (después de fusionar)
```bash
git branch -d nombre-de-la-rama        # Local
git push origin --delete nombre-rama   # Remota
```

### Flujo de Trabajo con Ramas (Git Flow Simplificado)
```
main ─────●─────●─────●─────●─────●───────
           \         /
            \───●───/  (feature/login)
```

1. Crear rama desde `main`: `git checkout -b feature/login`
2. Trabajar y hacer commits en esa rama
3. Fusionar de vuelta a `main` cuando esté listo
4. Eliminar la rama de trabajo

# Parte 3: GitHub Actions (Automatización)

Imagine que cada vez que realiza un `git push`, un servidor en la nube se activa, revisa el código y ejecuta tareas automáticamente. Eso define a GitHub Actions.

### Conceptos Fundamentales:
- **Workflow (Flujo de trabajo)**: El archivo de instrucciones completo.
- **Trigger (Disparador)**: El evento que activa el flujo (ej: al hacer `push`).
- **Job (Trabajo)**: Una serie de pasos agrupados.
- **Step (Paso)**: Una acción concreta (ej: "instalar python", "correr test").

### Primer Workflow: "Hola Mundo"
Crearemos un archivo que simplemente emitirá un saludo al subir código.

````python
%%writefile .github/workflows/hola_mundo.yml
name: Saludo Simple

# 1. El Disparador: Se activa al hacer push
on: [push]

jobs:
  # 2. El Trabajo
  saludar:
    runs-on: ubuntu-latest # Se utiliza un entorno Linux provisto por GitHub
    steps:
      - name: Obtener el código
        uses: actions/checkout@v4

      - name: Emitir saludo
        run: echo "¡Hola! El sistema funciona desde la nube de GitHub."
````

# Parte 4: Integración con Hugging Face

Hugging Face (HF) se consolida como el "GitHub de la IA". Permite alojar modelos, datasets y Demos (Spaces).

### El Flujo de Trabajo Ideal
1. Se trabaja en el código en **GitHub**.
2. Se realiza `git push`.
3. **GitHub Actions** copia automáticamente el código al **Space de Hugging Face**.

¡De esta forma se mantiene todo sincronizado eficientemente!

### Procedimiento de Configuración:

#### Paso 1: En Hugging Face
1. Cree un Space nuevo en [huggingface.co/new-space](https://huggingface.co/new-space).
2. Diríjase a **Settings → Access Tokens** y genere un token con permisos de escritura (`write`).
3. Anote su nombre de usuario de HF y el nombre del Space.

#### Paso 2: En GitHub (Secretos y Variables)
Diríjase a su repositorio → **Settings → Secrets and variables → Actions**.

**Crear un Secreto** (pestaña "Secrets"):
| Nombre | Valor |
|--------|-------|
| `HF_TOKEN` | Su token de Hugging Face (el que tiene permisos `write`) |

**Crear Variables** (pestaña "Variables"):
| Nombre | Valor | Ejemplo |
|--------|-------|---------|
| `HF_USERNAME` | Su nombre de usuario de HF | `mi-usuario` |
| `HF_SPACE_NAME` | Nombre del Space (sin el usuario) | `mi-aplicacion` |

> 💡 **Nota**: Los secretos se ocultan en los logs. Las variables son visibles pero no contienen información sensible.

#### Paso 3: El Workflow de Sincronización
Copie el siguiente código. Este proceso validará la configuración, tomará el código y lo enviará a Hugging Face con soporte para archivos grandes (Git LFS).

````python
%%writefile .github/workflows/sync_to_huggingface.yml
name: Sincronizar con Hugging Face

on:
  push:
    branches: [main]
  workflow_dispatch: # Permite ejecución manual desde GitHub

jobs:
  sync-to-hub:
    runs-on: ubuntu-latest
    steps:
      - name: Checkout del código
        uses: actions/checkout@v4
        with:
          fetch-depth: 0 # Importante: se descarga todo el historial
          lfs: true      # Soporte para archivos grandes (modelos)

      - name: Configurar Git
        run: |
          git config --global user.email "github-actions[bot]@users.noreply.github.com"
          git config --global user.name "github-actions[bot]"

      - name: Empujar a Hugging Face Hub
        env:
          HF_TOKEN: ${{ secrets.HF_TOKEN }}
          HF_USERNAME: ${{ vars.HF_USERNAME }}        # Configurar en Settings > Variables
          HF_SPACE_NAME: ${{ vars.HF_SPACE_NAME }}    # Nombre del Space (sin el usuario)
        run: |
          # Validar que las variables estén configuradas
          if [ -z "$HF_TOKEN" ]; then
            echo "❌ Error: HF_TOKEN no está configurado"
            exit 1
          fi
          if [ -z "$HF_USERNAME" ] || [ -z "$HF_SPACE_NAME" ]; then
            echo "❌ Error: HF_USERNAME o HF_SPACE_NAME no están configurados"
            echo "Configure las variables en: Settings > Secrets and variables > Actions > Variables"
            exit 1
          fi
          
          # Agregar el remoto de Hugging Face y hacer push
          git remote add huggingface https://$HF_USERNAME:$HF_TOKEN@huggingface.co/spaces/$HF_USERNAME/$HF_SPACE_NAME
          git push huggingface main --force
          
          echo "✅ Sincronización completada: https://huggingface.co/spaces/$HF_USERNAME/$HF_SPACE_NAME"
````

# Parte 5: Solución de Problemas Comunes (Troubleshooting)

A continuación se presentan los errores más frecuentes y sus soluciones.

---

### 5.1 Conflictos de Merge

**Síntoma**: Al hacer `git pull` o `git merge` aparece:
```
CONFLICT (content): Merge conflict in archivo.py
Automatic merge failed; fix conflicts and then commit the result.
```

**Solución**:
1. Abra el archivo en conflicto. Verá marcadores como:
```
<<<<<<< HEAD
su código local
=======
código remoto
>>>>>>> origin/main
```
2. Edite el archivo eliminando los marcadores y dejando el código correcto.
3. Guarde y ejecute:
```bash
git add archivo.py
git commit -m "Resolver conflicto en archivo.py"
```

---

### 5.2 Autenticación Fallida

**Síntoma**: `fatal: Authentication failed`

**Soluciones**:
- **Token expirado**: Genere un nuevo Personal Access Token en GitHub.
- **Credenciales cacheadas incorrectas**:
  ```bash
  # Windows
  git credential reject
  # macOS
  git credential-osxkeychain erase
  # Linux
  rm ~/.git-credentials
  ```
- Luego intente el `push` nuevamente e ingrese el nuevo token.

---

### 5.3 Push Rechazado (Non-Fast-Forward)

**Síntoma**: 
```
! [rejected] main -> main (non-fast-forward)
hint: Updates were rejected because the tip of your current branch is behind
```

**Causa**: Alguien más subió cambios mientras usted trabajaba.

**Solución**:
```bash
git pull origin main --rebase  # Integrar cambios remotos
git push origin main           # Intentar nuevamente
```

---

### 5.4 Archivos Grandes Rechazados

**Síntoma**: `File is too large` o errores de Hugging Face con modelos.

**Solución**: Instalar y usar Git LFS (Large File Storage):
```bash
git lfs install
git lfs track "*.bin" "*.h5" "*.safetensors"  # Extensiones de modelos
git add .gitattributes
git commit -m "Configurar Git LFS"
```

---

### 5.5 Workflow de GitHub Actions Falla

**Diagnóstico**:
1. Vaya a su repo → pestaña **Actions**.
2. Haga clic en el workflow fallido.
3. Expanda el paso que falló para ver el error.

**Errores comunes**:
| Error | Causa | Solución |
|-------|-------|----------|
| `HF_TOKEN not configured` | Falta el secreto | Agregar en Settings → Secrets |
| `Permission denied` | Token sin permisos `write` | Regenerar token con permisos correctos |
| `Repository not found` | Nombre de Space incorrecto | Verificar `HF_USERNAME` y `HF_SPACE_NAME` |

---

### 5.6 "Detached HEAD" 

**Síntoma**: `You are in 'detached HEAD' state`

**Causa**: Se hizo checkout a un commit específico en lugar de una rama.

**Solución**:
```bash
git checkout main  # Volver a la rama principal
# O si desea conservar cambios:
git checkout -b mi-rama-nueva
```
