## Introducción

En esta guía vamos a aprender a instalar Docker y n8n en tu computadora. Esta configuración te va a permitir trabajar con workflows de automatización de manera local y segura.

**Tiempo estimado:** 30-45 minutos  
**Nivel:** Principiante

---

## 1. ¿Qué es Docker y para qué se usa?

### Definición

Docker es una plataforma de software que permite **crear, distribuir y ejecutar aplicaciones dentro de contenedores**. Podemos pensar en un contenedor como una "caja" que contiene todo lo necesario para que una aplicación funcione: el código, las librerías, las dependencias y las configuraciones.

### ¿Por qué usamos Docker?

Docker resuelve varios problemas comunes en el desarrollo de software:

1. **Portabilidad:** Una aplicación que funciona en tu computadora va a funcionar igual en cualquier otra que tenga Docker instalado.

2. **Aislamiento:** Cada contenedor funciona de manera independiente, sin interferir con otras aplicaciones de tu sistema.

3. **Eficiencia:** Los contenedores comparten el núcleo del sistema operativo, haciéndolos mucho más livianos que las máquinas virtuales tradicionales.

4. **Reproducibilidad:** Todos los estudiantes van a trabajar con el mismo entorno, eliminando el clásico problema de "en mi máquina funciona".

### Conceptos clave

- **Imagen:** Es como una "plantilla" o "receta" que define qué contiene el contenedor.
- **Contenedor:** Es una instancia en ejecución de una imagen.
- **Docker Hub:** Repositorio público donde se almacenan imágenes de Docker que podemos descargar y usar.
- **Volumen:** Espacio de almacenamiento persistente que sobrevive incluso si el contenedor se detiene o elimina.

### ¿Por qué usamos Docker para n8n?

n8n es una herramienta de automatización de workflows que requiere Node.js y varias dependencias. En lugar de instalar todo esto manualmente en cada computadora, usamos Docker para:

- Instalar n8n con un solo comando
- Mantener nuestro sistema limpio y organizado
- Actualizar n8n fácilmente cuando salgan nuevas versiones
- Garantizar que todos trabajamos con la misma configuración

## 2. Requisitos previos

Antes de comenzar, asegurate de tener:

### Hardware mínimo
- **RAM:** 4 GB mínimo (recomendado 8 GB)
- **Disco:** 10 GB de espacio libre
- **Procesador:** Compatible con virtualización (la mayoría de las computadoras modernas lo soportan)

### Software
- **Sistema operativo:** Windows 10/11, macOS 10.15+, o Linux (Ubuntu 20.04+)
- **Permisos:** Derechos de administrador en tu computadora
- **Navegador web:** Chrome, Firefox, Edge o Safari actualizado

### Conocimientos previos
- Uso básico de la terminal o línea de comandos
- Navegación por el sistema de archivos

## 3. Instalación de Docker

### 3.1. Instalación según tu sistema operativo

#### Windows

1. **Descargá Docker Desktop:**
   - Visitá: https://www.docker.com/products/docker-desktop
   - Descargá la versión para Windows

2. **Instalá Docker Desktop:**
   - Ejecutá el instalador descargado
   - Seguí las instrucciones del asistente
   - Reiniciá tu computadora cuando te lo pida

3. **Configuración inicial:**
   - Abrí Docker Desktop
   - Aceptá los términos de servicio
   - Esperá a que Docker inicie (puede tardar unos minutos la primera vez)

#### macOS

1. **Descargá Docker Desktop:**
   - Visitá: https://www.docker.com/products/docker-desktop
   - Seleccioná la versión correcta:
     - **Mac con Intel:** Descargá la versión Intel
     - **Mac con M1/M2/M3:** Descargá la versión Apple Silicon

2. **Instalá Docker Desktop:**
   - Abrí el archivo .dmg descargado
   - Arrastrá Docker a la carpeta Aplicaciones
   - Abrí Docker desde Aplicaciones

3. **Permisos:**
   - Ingresá tu contraseña cuando te lo pida
   - Esperá a que Docker inicie

#### Linux (Ubuntu/Debian)

Ejecutá estos comandos en la terminal:

```bash
# Actualizá el sistema
sudo apt update
sudo apt upgrade -y

# Instalá dependencias necesarias
sudo apt install -y apt-transport-https ca-certificates curl software-properties-common

# Agregá el repositorio oficial de Docker
curl -fsSL https://download.docker.com/linux/ubuntu/gpg | sudo gpg --dearmor -o /usr/share/keyrings/docker-archive-keyring.gpg

echo "deb [arch=$(dpkg --print-architecture) signed-by=/usr/share/keyrings/docker-archive-keyring.gpg] https://download.docker.com/linux/ubuntu $(lsb_release -cs) stable" | sudo tee /etc/apt/sources.list.d/docker.list > /dev/null

# Instalá Docker
sudo apt update
sudo apt install -y docker-ce docker-ce-cli containerd.io

# Agregá tu usuario al grupo docker (para no usar sudo)
sudo usermod -aG docker $USER

# Cerrá sesión y volvé a iniciar para que los cambios tengan efecto
```

## 4. Verificación de la instalación de Docker

Después de instalar Docker, necesitamos verificar que funcione correctamente.

### 4.1. Verificá que Docker esté instalado

Abrí una terminal (Windows: PowerShell o CMD, macOS/Linux: Terminal) y ejecutá:

```bash
docker --version
```

**Resultado esperado:** Deberías ver algo como:
```
Docker version 24.0.6, build ed223bc
```

### 4.2. Verificá que Docker esté en ejecución

```bash
docker ps
```

**Resultado esperado:** Una tabla vacía (por ahora no tenemos contenedores corriendo):
```
CONTAINER ID   IMAGE     COMMAND   CREATED   STATUS    PORTS     NAMES
```

### 4.3. Probá Docker con un contenedor de prueba

```bash
docker run hello-world
```

**Resultado esperado:** Un mensaje de bienvenida que confirma que Docker está funcionando correctamente.

### Solución de problemas comunes

**Error: "Docker daemon is not running"**
- **Windows/macOS:** Abrí Docker Desktop y esperá a que inicie
- **Linux:** Ejecutá `sudo systemctl start docker`

**Error: "Permission denied"**
- **Linux:** Cerrá sesión y volvé a iniciar después de agregar tu usuario al grupo docker
- O ejecutá los comandos con `sudo` (no recomendado para uso regular)

## 5. ¿Qué es n8n?

### Definición

n8n (pronunciado "n-eight-n") es una herramienta de **automatización de workflows** de código abierto. Te permite conectar diferentes aplicaciones y servicios para automatizar tareas repetitivas sin necesidad de programar.

### ¿Para qué se usa n8n?

n8n se utiliza para:

1. **Integrar servicios:** Conectar aplicaciones como Gmail, Slack, Google Sheets, bases de datos, APIs, etc.

2. **Automatizar procesos:** Crear flujos de trabajo que se ejecutan automáticamente según disparadores (triggers) específicos.

3. **Procesar datos:** Transformar, filtrar y manipular información entre diferentes sistemas.

4. **Crear soluciones personalizadas:** Desarrollar integraciones específicas para necesidades particulares.

### Ejemplos de uso

- Guardar automáticamente adjuntos de emails en Google Drive
- Enviar notificaciones a Slack cuando se actualiza una base de datos
- Generar reportes automáticos combinando datos de múltiples fuentes
- Sincronizar información entre sistemas diferentes

### Ventajas de n8n

- **Open source:** Código abierto y gratuito
- **Self-hosted:** Controlás tus datos al ejecutarlo en tu propia infraestructura
- **Interfaz visual:** No necesitás ser programador para crear workflows
- **Extensible:** Podés crear tus propios nodos personalizados
- **400+ integraciones:** Amplia biblioteca de conectores preconfigurados

## 6. Instalación de n8n con Docker

Ahora que tenemos Docker funcionando, vamos a instalar n8n. Existen dos formas principales de ejecutar n8n:

1. **Modo temporal:** Para pruebas rápidas (los datos se pierden al cerrar)
2. **Modo persistente:** Para uso regular (los datos se guardan permanentemente)

### 6.1. Instalación rápida (modo temporal)

Esta opción es útil para probar n8n por primera vez. Los workflows y configuraciones se perderán al detener el contenedor.

```bash
docker run -it --rm --name n8n -p 5678:5678 n8nio/n8n
```

**Explicación del comando:**
- `docker run`: Ejecuta un contenedor
- `-it`: Modo interactivo con terminal
- `--rm`: Elimina el contenedor automáticamente al detenerlo
- `--name n8n`: Asigna el nombre "n8n" al contenedor
- `-p 5678:5678`: Mapea el puerto 5678 del contenedor al puerto 5678 de tu computadora
- `n8nio/n8n`: Imagen oficial de n8n en Docker Hub

**Acceso:** Abrí tu navegador y visitá: http://localhost:5678

**Para detener n8n:** Presioná `Ctrl + C` en la terminal

---

### 6.2. Instalación con persistencia de datos (recomendado)

Esta es la configuración que vamos a usar durante el curso. Los workflows, credenciales y configuraciones se guardan en tu computadora y persisten entre reinicios.

#### Paso 1: Creá un directorio para n8n

**Windows (PowerShell):**
```powershell
mkdir C:\n8n-data
cd C:\n8n-data
```

**macOS/Linux:**
```bash
mkdir -p ~/.n8n
```

#### Paso 2: Ejecutá n8n con volumen persistente

**Windows (PowerShell):**
```powershell
docker run -d `
  --name n8n `
  -p 5678:5678 `
  -v C:\n8n-data:/home/node/.n8n `
  --restart unless-stopped `
  n8nio/n8n
```

**macOS/Linux:**
```bash
docker run -d \
  --name n8n \
  -p 5678:5678 \
  -v ~/.n8n:/home/node/.n8n \
  --restart unless-stopped \
  n8nio/n8n
```

**Explicación del comando:**
- `-d`: Ejecuta el contenedor en segundo plano (detached mode)
- `-v`: Monta un volumen para persistir datos
  - **Windows:** `C:\n8n-data` se mapea a `/home/node/.n8n` dentro del contenedor
  - **macOS/Linux:** `~/.n8n` (carpeta oculta en tu home) se mapea al contenedor
- `--restart unless-stopped`: Reinicia automáticamente el contenedor si se detiene (excepto si lo detuviste manualmente)

#### Paso 3: Verificá que n8n esté corriendo

```bash
docker ps
```

Deberías ver algo como:
```
CONTAINER ID   IMAGE        COMMAND                  STATUS         PORTS                    NAMES
abc123def456   n8nio/n8n    "tini -- /docker-ent…"   Up 2 minutes   0.0.0.0:5678->5678/tcp   n8n
```

#### Paso 4: Accedé a n8n

Abrí tu navegador y visitá: **http://localhost:5678**

La primera vez que accedas, n8n te va a pedir que crees una cuenta de administrador.

## 7. Configuración inicial de n8n

### 7.1. Primera vez en n8n

Cuando accedas a http://localhost:5678 por primera vez, vas a ver una pantalla de bienvenida.

#### Creá tu cuenta de administrador

1. **Email:** Ingresá tu email universitario (ej: `tunombre@estudiante.uba.ar`)
2. **Contraseña:** Elegí una contraseña segura
3. **Nombre:** Tu nombre completo

**Importante:** Guardá estas credenciales en un lugar seguro. Las vas a necesitar cada vez que accedas a n8n.

### 7.2. Exploración de la interfaz

Una vez dentro, familiarizate con los elementos principales:

1. **Workflows:** Lista de tus flujos de trabajo
2. **Credentials:** Donde guardás las credenciales de APIs y servicios
3. **Executions:** Historial de ejecuciones de workflows
4. **Settings:** Configuración general de n8n

### 7.3. Creá tu primer workflow de prueba

1. Hacé clic en "New Workflow" (o "Nuevo Workflow")
2. Aparecerá el editor visual con un canvas en blanco
3. Hacé clic en el botón "+" para agregar un nodo
4. Buscá y agregá un nodo "Manual Trigger" (Disparador Manual)
5. Agregá otro nodo, por ejemplo "Set" para manipular datos
6. Conectá los nodos arrastrando desde los conectores
7. Guardá el workflow con un nombre descriptivo

Este es solo un ejercicio de familiarización. En las próximas clases vamos a ver en detalle cómo crear workflows más complejos.

## 8. Comandos útiles de Docker para n8n

### Ver contenedores en ejecución

```bash
docker ps
```

### Ver todos los contenedores (incluidos los detenidos)

```bash
docker ps -a
```

### Detener n8n

```bash
docker stop n8n
```

### Iniciar n8n (si está detenido)

```bash
docker start n8n
```

### Reiniciar n8n

```bash
docker restart n8n
```

### Ver los logs de n8n (útil para debugging)

```bash
docker logs n8n
```

### Ver logs en tiempo real

```bash
docker logs -f n8n
```

Presioná `Ctrl + C` para salir de los logs.

### Eliminar el contenedor de n8n

**Atención:** Esto elimina el contenedor, pero NO los datos (si usaste volúmenes persistentes).

```bash
docker stop n8n
docker rm n8n
```

### Actualizar n8n a la última versión

```bash
# Detener el contenedor actual
docker stop n8n

# Eliminar el contenedor actual
docker rm n8n

# Descargar la última versión de la imagen
docker pull n8nio/n8n

# Volver a crear el contenedor con la nueva versión
# (usar el mismo comando del paso 6.2)
```

### Acceder a la terminal del contenedor

```bash
docker exec -it n8n sh
```

Para salir de la terminal del contenedor, escribí `exit`.

## 9. Solución de problemas comunes

### Problema 1: No puedo acceder a http://localhost:5678

**Soluciones:**

1. **Verificá que el contenedor esté corriendo:**
   ```bash
   docker ps
   ```
   Si no aparece n8n, inicialo con `docker start n8n`

2. **Verificá que el puerto 5678 esté disponible:**
   - Puede que otra aplicación esté usando ese puerto
   - Probá cambiar el puerto en el comando docker run: `-p 5679:5678`
   - Luego accedé a http://localhost:5679

3. **Revisá los logs:**
   ```bash
   docker logs n8n
   ```

### Problema 2: El contenedor se detiene inmediatamente

**Soluciones:**

1. **Revisá los logs para ver el error:**
   ```bash
   docker logs n8n
   ```

2. **Verificá los permisos del volumen:**
   - Asegurate de que el directorio de datos tenga permisos de escritura

### Problema 3: Los workflows desaparecen al reiniciar

**Causa:** No configuraste correctamente el volumen persistente.

**Solución:** Eliminá el contenedor y volvé a crearlo con el comando del paso 6.2 que incluye el parámetro `-v`.

### Problema 4: Error "port is already allocated"

**Causa:** Otro servicio está usando el puerto 5678.

**Soluciones:**

1. **Opción A:** Detené el servicio que usa ese puerto

2. **Opción B:** Usá otro puerto para n8n:
   ```bash
   docker run -d --name n8n -p 8080:5678 -v ~/.n8n:/home/node/.n8n n8nio/n8n
   ```
   Luego accedé a http://localhost:8080

### Problema 5: Docker Desktop no inicia (Windows/macOS)

**Soluciones:**

1. **Windows:**
   - Verificá que la virtualización esté habilitada en la BIOS
   - Asegurate de tener WSL 2 instalado
   - Reiniciá tu computadora

2. **macOS:**
   - Verificá que tengas suficiente espacio en disco
   - Intentá reinstalar Docker Desktop
   - Revisá los logs de sistema en Console.app

### Problema 6: Rendimiento lento

**Soluciones:**

1. **Aumentá los recursos de Docker Desktop:**
   - Abrí Docker Desktop > Settings > Resources
   - Aumentá la RAM asignada (recomendado: 4-6 GB)
   - Aumentá los CPUs asignados

2. **Cerrá otras aplicaciones pesadas** mientras trabajás con n8n

## 10. Variables de entorno útiles (opcional)

Podés configurar n8n con variables de entorno para personalizar su comportamiento. Estas se agregan al comando `docker run` con el parámetro `-e`.

### Ejemplo con variables de entorno

```bash
docker run -d \
  --name n8n \
  -p 5678:5678 \
  -e N8N_BASIC_AUTH_ACTIVE=true \
  -e N8N_BASIC_AUTH_USER=estudiante \
  -e N8N_BASIC_AUTH_PASSWORD=password123 \
  -e GENERIC_TIMEZONE=America/Argentina/Buenos_Aires \
  -v ~/.n8n:/home/node/.n8n \
  n8nio/n8n
```

### Variables útiles

| Variable | Descripción | Ejemplo |
|----------|-------------|----------|
| `GENERIC_TIMEZONE` | Zona horaria | `America/Argentina/Buenos_Aires` |
| `N8N_BASIC_AUTH_ACTIVE` | Activar autenticación básica | `true` |
| `N8N_BASIC_AUTH_USER` | Usuario para auth básica | `estudiante` |
| `N8N_BASIC_AUTH_PASSWORD` | Contraseña para auth básica | `password123` |
| `N8N_PORT` | Puerto interno (no lo cambies) | `5678` |
| `WEBHOOK_URL` | URL base para webhooks | `http://localhost:5678/` |
| `N8N_LOG_LEVEL` | Nivel de logging | `info`, `debug`, `error` |

**Nota:** Para uso local en el curso, las configuraciones por defecto son suficientes.
